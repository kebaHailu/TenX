```python
# Import Dataset and useful libraries 
import os 
import pandas as pd 
os.chdir('../scripts/')
import utils as util 


data_path = "../../data/week2/source.csv"
df = util.read_csv_file(data_path)
data = df.get("data")
```

```python
# Handling missing values and outliers
data['TCP DL Retrans. Vol (Bytes)'].fillna(data['TCP DL Retrans. Vol (Bytes)'].mean(), inplace=True)
data['Avg RTT DL (ms)'].fillna(data['Avg RTT DL (ms)'].mean(), inplace=True)
data['Avg Bearer TP DL (kbps)'].fillna(data['Avg Bearer TP DL (kbps)'].mean(), inplace=True)

# Assuming 'Handset Type' is categorical, we can fill missing values with the mode
data['Handset Type'].fillna(data['Handset Type'].mode()[0], inplace=True)

# Aggregating the data per customer
aggregated_data = data.groupby('Bearer Id').agg(
    Average_TCP_retransmission=('TCP DL Retrans. Vol (Bytes)', 'mean'),
    Average_RTT=('Avg RTT DL (ms)', 'mean'),
    Handset_type=('Handset Type', 'first'),  # Assuming we want the first occurrence
    Average_Throughput=('Avg Bearer TP DL (kbps)', 'mean')
).reset_index()

# Display the aggregated data
print(aggregated_data)
```

```python
# Top 10 TCP values
top_tcp_values = data['TCP DL Retrans. Vol (Bytes)'].nlargest(10)
print("Top 10 TCP values:")
print(top_tcp_values)

# Bottom 10 TCP values
bottom_tcp_values = data['TCP DL Retrans. Vol (Bytes)'].nsmallest(10)
print("\nBottom 10 TCP values:")
print(bottom_tcp_values)

# Most frequent TCP values
most_frequent_tcp_values = data['TCP DL Retrans. Vol (Bytes)'].value_counts().nlargest(10)
print("\nMost frequent TCP values:")
print(most_frequent_tcp_values)

# Top 10 RTT values
top_rtt_values = data['Avg RTT DL (ms)'].nlargest(10)
print("\nTop 10 RTT values:")
print(top_rtt_values)

# Bottom 10 RTT values
bottom_rtt_values = data['Avg RTT DL (ms)'].nsmallest(10)
print("\nBottom 10 RTT values:")
print(bottom_rtt_values)

# Most frequent RTT values
most_frequent_rtt_values = data['Avg RTT DL (ms)'].value_counts().nlargest(10)
print("\nMost frequent RTT values:")
print(most_frequent_rtt_values)

# Top 10 Throughput values
top_throughput_values = data['Avg Bearer TP DL (kbps)'].nlargest(10)
print("\nTop 10 Throughput values:")
print(top_throughput_values)

# Bottom 10 Throughput values
bottom_throughput_values = data['Avg Bearer TP DL (kbps)'].nsmallest(10)
print("\nBottom 10 Throughput values:")
print(bottom_throughput_values)

# Most frequent Throughput values
most_frequent_throughput_values = data['Avg Bearer TP DL (kbps)'].value_counts().nlargest(10)
print("\nMost frequent Throughput values:")
print(most_frequent_throughput_values)
```

```python
# Average throughput per handset type
average_throughput_per_handset = data.groupby('Handset Type')['Avg Bearer TP DL (kbps)'].mean()
print("\nAverage Throughput per Handset Type:")
print(average_throughput_per_handset)

# Interpretation of average throughput findings
print("\nInterpretation of Average Throughput Findings:")
for handset_type, throughput in average_throughput_per_handset.items():
    print(f"{handset_type}: {throughput:.2f} kbps - This indicates the average data throughput experienced by users with {handset_type}.")

# Average TCP retransmission per handset type
average_tcp_retransmission_per_handset = data.groupby('Handset Type')['TCP DL Retrans. Vol (Bytes)'].mean()
print("\nAverage TCP Retransmission per Handset Type:")
print(average_tcp_retransmission_per_handset)

# Interpretation of average TCP retransmission findings
print("\nInterpretation of Average TCP Retransmission Findings:")
for handset_type, retransmission in average_tcp_retransmission_per_handset.items():
    print(f"{handset_type}: {retransmission:.2f} Bytes - This indicates the average volume of TCP retransmissions for users with {handset_type}.")
```

```python
from sklearn.cluster import KMeans
import pandas as pd

# Selecting relevant features for clustering
features = data[['Avg RTT DL (ms)', 'Avg Bearer TP DL (kbps)', 'TCP DL Retrans. Vol (Bytes)']]

# Performing k-means clustering
kmeans = KMeans(n_clusters=3, random_state=42)
data['Cluster'] = kmeans.fit_predict(features)

# Describing each cluster
cluster_description = data.groupby('Cluster').agg({
    'Avg RTT DL (ms)': 'mean',
    'Avg Bearer TP DL (kbps)': 'mean',
    'TCP DL Retrans. Vol (Bytes)': 'mean',
    'Handset Type': lambda x: x.mode()[0]  # Most common handset type in the cluster
}).reset_index()

print("\nCluster Descriptions:")
for index, row in cluster_description.iterrows():
    print(f"Cluster {row['Cluster']}:")
    print(f"  Average RTT: {row['Avg RTT DL (ms)']:.2f} ms")
    print(f"  Average Throughput: {row['Avg Bearer TP DL (kbps)']:.2f} kbps")
    print(f"  Average TCP Retransmission: {row['TCP DL Retrans. Vol (Bytes)']:.2f} Bytes")
    print(f"  Most Common Handset Type: {row['Handset Type']}")
    print()
```
